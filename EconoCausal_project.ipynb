```python

# 1. Duplicate checks
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())

# 2. Treatment and outcome values
print("\nDiscount distribution:")
print(df["discount_offered"].value_counts(dropna=False).sort_index())

print("\nPurchase distribution:")
print(df["purchased"].value_counts(dropna=False).sort_index())

# 3. Binary outcome validation
print("\nInvalid purchased values:",
      (~df["purchased"].isin([0, 1])).sum())

# 4. Missing values (all columns)
print("\nColumns with missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# 5. Numeric columns with negative values
numeric_df = df.select_dtypes(include="number")
print("\nNegative values by numeric column:")
print((numeric_df < 0).sum()[(numeric_df < 0).sum() > 0])

# 6. Economic columns
economic_cols = [
    "discount_amount", "revenue", "cost_of_goods",
    "marketing_cost", "net_profit"
]
print("\nEconomic columns summary:")
print(df[economic_cols].describe().round(2))

```

```python

# Profit ko purchase status ke according compare karo
print(
    df.groupby("purchased")["net_profit"]
      .agg(["count", "mean", "min", "median", "max"])
      .round(2)
)

# Purchase status aur net-profit sign ka relationship
print("\nPurchase status vs profit sign:")
print(
    pd.crosstab(
        df["purchased"],
        df["net_profit"] < 0,
        margins=True
    )
)

# Zero-revenue rows ka profit
print("\nRows with zero revenue:")
print("Count:", (df["revenue"] == 0).sum())
print(
    df.loc[df["revenue"] == 0,
           ["purchased", "revenue", "cost_of_goods",
            "marketing_cost", "net_profit"]]
      .head(10)
)
```

```python

# Check 1: Net profit formula
expected_profit = (
    df["revenue"]
    - df["cost_of_goods"]
    - df["marketing_cost"]
)

profit_diff = (df["net_profit"] - expected_profit).abs()

print("Net profit formula mismatches:", (profit_diff > 0.02).sum())
print("Maximum absolute difference:", round(profit_diff.max(), 4))


# Check 2: Revenue formula for purchased customers
buyers = df["purchased"] == 1

expected_revenue = (
    df.loc[buyers, "basket_value_before_discount"]
    - df.loc[buyers, "discount_amount"]
)

revenue_diff = (
    df.loc[buyers, "revenue"] - expected_revenue
).abs()

print("\nBuyer revenue mismatches:", (revenue_diff > 0.02).sum())
print("Maximum revenue difference:", round(revenue_diff.max(), 4))


# Check 3: Negative-profit purchasers
negative_buyers = df[
    (df["purchased"] == 1) & (df["net_profit"] < 0)
]

print("\nPurchasers with negative profit:", len(negative_buyers))
print(
    negative_buyers[
        ["discount_offered", "discount_amount",
         "basket_value_before_discount", "revenue",
         "cost_of_goods", "marketing_cost", "net_profit"]
    ].head(10).to_string(index=False)
)
```

```python

# Purchase rate for each discount level
purchase_summary = (
    df.groupby("discount_offered")
      .agg(
          customers=("customer_id", "count"),
          purchases=("purchased", "sum"),
          purchase_rate=("purchased", "mean"),
          avg_revenue=("revenue", "mean"),
          avg_profit=("net_profit", "mean")
      )
)

purchase_summary["purchase_rate_pct"] = (
    purchase_summary["purchase_rate"] * 100
).round(2)

print(purchase_summary.round(3))

# Discount levels should be exactly the expected categories
print("\nDiscount levels:", sorted(df["discount_offered"].unique()))
print("Purchase values:", sorted(df["purchased"].unique()))
```

```python

# Check relationship between discount and marketing costs
print(
    df.groupby("discount_offered")[
        ["discount_amount", "marketing_cost", "revenue",
         "cost_of_goods", "net_profit"]
    ].mean().round(2)
)

print("\nMarketing cost minus discount amount:")
print(
    (df["marketing_cost"] - df["discount_amount"])
    .describe()
    .round(2)
)

print("\nCorrelation between discount and marketing cost:")
print(
    df[["discount_amount", "marketing_cost"]]
    .corr()
    .round(3)
)
```

```python

# Check the exact accounting identity
df["calculated_profit"] = (
    df["revenue"]
    - df["cost_of_goods"]
    - df["marketing_cost"]
)

df["profit_difference"] = (
    df["net_profit"] - df["calculated_profit"]
)

print("Formula mismatches:",
      (df["profit_difference"].abs() > 0.01).sum())

print("\nSample rows:")
print(
    df[[
        "discount_offered",
        "discount_amount",
        "revenue",
        "cost_of_goods",
        "marketing_cost",
        "net_profit",
        "calculated_profit",
        "profit_difference"
    ]].head(10).round(2)
)
```

```python

profit_breakdown = (
    df.groupby(["discount_offered", "purchased"])
      .agg(
          customers=("customer_id", "count"),
          avg_revenue=("revenue", "mean"),
          avg_cost_of_goods=("cost_of_goods", "mean"),
          avg_marketing_cost=("marketing_cost", "mean"),
          avg_net_profit=("net_profit", "mean")
      )
      .round(2)
)

print(profit_breakdown)
```

```python

import json

with open("EconoCausal_project.ipynb", "r", encoding="utf-8") as f:
    notebook = json.load(f)

keywords = [
    "discount_offered",
    "discount_amount",
    "marketing_cost",
    "net_profit",
    "recency_days",
    "customer_satisfaction",
]

for i, cell in enumerate(notebook["cells"]):
    source = "".join(cell.get("source", []))
    if cell["cell_type"] == "code" and any(
        keyword in source for keyword in keywords
    ):
        print(f"\n--- Cell {i} ---")
        print(source[:2500])
```
